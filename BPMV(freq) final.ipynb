# I. Identifying Business Needs
The online car resale company Cars 4 You had successfully accumulated a list of customers due to its useful business model. The platform allowed people who wish to sell their cars to provide information regarding the vehicle and send it to their chain of mechanics so that it can be assessed before being purchased and then resold. Unfortunately, with the fame and company growth, waiting lists of car evaluation had significantly enlarged. 

To solve the issue, it was developed a predictive machine learning model which estimates the cars' value based on the inputs provided by the user. This way, it will not be necessary to take the car to a mechanic, hence the company will become capable to handle bigger number of customers and to be more efficient 

Several process and methods were use during the creation of the predictive model. First, two datasets were obtained from a database in 2020, train_data and test_data.
```python
#checking outliers of numerical variables through the visualization of boxplots

def plot_multiple_boxplots(data, feats, title="Numeric Variables' Box Plots"):
    
    #creating a grid of subplots with 4 rows and as many columns
    #as needed to place all numerical features
    fig, axes = plt.subplots(
        4,
        ceil(len(feats) / 4),
        figsize=(40, 30)
    )
    
    #plotting data
    #iterating across axes objects and associating each histogram
    for ax, feat in zip(axes.flatten(), feats):
        sns.boxplot(
            x=data[feat],        #data for the current feature
            ax=ax,               #axis where the plot is drawn
            color="#5dade2"      #boxplot color
        )
        ax.set_title(feat)      #setting subplot title as feature name

    #removing unused subplots: if the total number of features does not fill the grid,
    # remove the empty axes 
    axes_flat = axes.flatten()
    for i in range(len(feats), len(axes_flat)):
        fig.delaxes(axes_flat[i])

    #adding a title and displaying the figure
    (plt.suptitle(title, fontsize=30, y=0.92))
    plt.show()

    return
    
'''This function visualizes the distribution and potential outliers of
multiple numerical variables using boxplots. Given a dataset and a list
of numerical features, it automatically creates a grid of boxplots,
where each subplot corresponds to one feature. '''
```

```python
#creating correlation matrix to facilitate interpretation
def cor_heatmap(cor):
    
    #setting the figure size
    plt.figure(figsize=(12, 10))

    #creating a mask for the upper triangle of the matrix (to avoid plotting duplicate correlation values)
    mask = np.triu(np.ones_like(cor, dtype=bool))

    #plotting the correlation heatmap
    sns.heatmap(
        data=cor,                 #correlation matrix input
        mask=mask,                #applying upper-triangle mask
        annot=True,               #displaying correlation coefficients
        cmap='YlGnBu',            #color map for visual contrast
        fmt='.2f',                #formatting values to two decimals
        square=True,              #ensuring square-shaped cells
        linewidths=0.5,           #adding grid lines between cells
        cbar_kws={"shrink": 0.8}  #adjusting color bar size
    )

    #adding a title and display the plot
    plt.title("Spearman Correlation Matrix", fontsize=14)
    plt.show()

'''This function visualises a correlation matrix using a heatmap to
facilitate interpretation of relationships between variables. By 
masking the upper triangle of the matrix, the plot avoids redundant
information and improves the readability. '''
```

### 4.2. Split the data into train and validation datasets

#### The Hold Out Method

In order to understand if a predictive model works, we need to be able to evaluate how it performs. Therefore, we are going to split the dataset (*data*) into *train* and *validation* using *The Hold Out Method* because not only this method it is efficient and simple to implement but also due to the nature of our dataset, that is characterized by being large and approximately i.i.d. 
To apply this method, we first seperated the price (target festure) from the original dataset and then split the remaining data into **X_val** (30% of *data*) and **X_train** (70% of *data*). From this point onward, both *X_train* and *X_val* will be modified in each stage, with preprocessing techniques being fit on *X_train* and then applied (transformed) to *X_val* to prevent data leakage. 

**Note** that the *test* dataset does not need to be splited. 

```python
X = data.drop('price', axis = 1) #in X, the target variable will be removed and the dataset will be used as the training set
y = data['price']  #y corresponds to the target variable
```

```python
#splitting the dataset

X_train, X_val, y_train, y_val = train_test_split(X,y, test_size = 0.3,  #30% will be used for validation 
                                                  random_state = 0,      #ensuring the split is always the same every time the code runs
                                                  shuffle = True)        #shuffling the data before spliting to avoid bias
```

### 4.3. Changing datatypes

It did not make sense for *year*, *previousOwners* and *hasDamage* to be floats, so we have decided to change their datatype to integer. Note that, we previously stated that *hasDamaged* should be a boolean, however; a feature cannot change its datatype to *bool* if it has missing values. In this way, it was changed to *int8*, where 0 corresponds to "False" and 1 corresponds to "True".

```python
#rounding the floats and changing them to integers

#year to integer
X_train['year'] = X_train['year'].round().astype('Int32')
X_val['year'] = X_val['year'].round().astype('Int32')
test_data['year'] = test_data['year'].round().astype('Int32')

#previousOwners to integer
X_train['previousOwners'] = X_train['previousOwners'].round().astype('Int32')
X_val['previousOwners'] = X_val['previousOwners'].round().astype('Int32')
test_data['previousOwners'] = test_data['previousOwners'].round().astype('Int32')

#hasDamaged to boolean
X_train['hasDamage'] = X_train['hasDamage'].astype('Int8')
X_val['hasDamage'] = X_val['hasDamage'].astype('Int8')
test_data['hasDamage'] = test_data['hasDamage'].astype('Int8')
```

### 4.7. Treating Missing Values
Considering that most of techniques used to handle incoherencies and treating outliers in numerical features consisted on replacing the numbers by missing values, we needed a good technique that was able to fill the missing values according to our dataset. Through the analysis made in the previous steps, we understood that if we want to fill a missing value of a certain variable, we cannot solely consider existent values of the same variable, we have to consider the rest of the variables of the dataset as well. For example, if our objective is to fill the *engineSize* of a certain observation, we should take into account its model and brand, assuming that is likely for a *Ford focus* to have the same or similiar engine size as other *Ford focus* that may exist in our dataset. 
For this reason, we have decided to fill numerical and categorical features using **Interactive Imputer** with **Extra Trees** as an estimator, which fill each variable with missing values using all the other variables as predictors. Moreover, *ExtraTrees* is abble to learn the nonlinear relationships present in our data and it captures interaction between variables. 

#### 4.7.1. Handling Categorical Features

The first step to apply the interactive imputer was transform the categorical features into numerical using *Frequency Encoding*. This action was required because the model that Iterative Imputer uses internally (in this case, *Extra Trees*) is a mathematical regression algorithm that can only operate on numeric features. We used *Frequency Encoding* because missing values remain missing during the process, it simply counts how frequent each category has appeared. 

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.ensemble import ExtraTreesRegressor, ExtraTreesClassifier
from sklearn.neural_network import MLPRegressor
```

#### 4.7.2. Interactive Imputer
After encoding the categorical variables of the datasets using *Frequency Encoding*, it was assumed that these variables were now numerical. Therefore, the estimator were are going to use to impute the missing values is *Extra Trees Regressor*, which is suitable for numerical varibles. In the case of using *One Hot Encoder*, for example, it would be necessaru to apply *Extra Trees Classifier* beacuse the variables would still have a categorical caracter. 

```python
#defining the columns
num_cols = ['mileage', 'mpg', 'tax', 'engineSize', 'previousOwners', 'year', 'fuelType_encoded', 'model_encoded', 'brand_encoded', 'transmission_encoded', 'hasDamage']
```

```python
#making a copy of the datasets
X_train_imp = X_train.copy()
X_val_imp   = X_val.copy()
test_imp    = test_data.copy()

#defining ExtraTrees regressor for numerical imputation
num_imputer_ETR = ExtraTreesRegressor(
    n_estimators=80,        #number of trees
    max_depth=12,           #maximum depth of each tree
    min_samples_leaf=2,     #minimum samples per leaf
    max_features=None,      #using all features at each split
    n_jobs=-1,              #parallel computation
    random_state=42         #reproducibility
)

#IterativeImputer models each feature with missing values
#as a function of the remain features 
num_imputer = IterativeImputer(
    estimator=num_imputer_ETR,    #estimator used for prediction
    max_iter=12,                  #number of imputation rounds
    initial_strategy='median',    #initial fill strategy
    imputation_order='ascending', #order of feature imputation
    skip_complete=True,           #skiping features without missing values
    random_state=42               #reproducibility
)

# fitting the imputer only on training data to avoid data leakage
train_imputed = num_imputer.fit_transform(X_train_imp[num_cols])

#applying imputation patterns on validation and test datasets
val_imputed   = num_imputer.transform(X_val_imp[num_cols])
test_imputed  = num_imputer.transform(test_imp[num_cols])
 
#replacing original numerical columns with imputed versions
X_train_imp[num_cols] = train_imputed
X_val_imp[num_cols]   = val_imputed
test_imp[num_cols]    = test_imputed

#checking if all missing values were handled
print("Missing after imputation (TRAIN):\n", X_train_imp[num_cols].isna().sum())
print("\nMissing after imputation (VAL):\n", X_val_imp[num_cols].isna().sum())
print("\nMissing after imputation (TEST):\n", test_imp[num_cols].isna().sum())

'''This code performs numerical missing value imputation using an 
Iterative Imputer with an ExtraTrees regressor as the predictive model. 
The imputer is trained exclusively on the training dataset to prevent
data leakage and then applied to the validation and test datasets.'''
```

```python
#applying the imputed datasets into the original ones
X_train = X_train_imp
X_val = X_val_imp
test_data = test_imp
```

### 5.2. Wrapper Methods 
In addition to the filter methods, we also analysed wrapper methods that evaluate features based on actual model performances. In this way, we were able to detect interactions between features, since some of them may be considered either weak alone but strong when combined, or redundant when another feature is present in the model. Here, the based estimator was *Random Forest Regressor* and we evaluated the sets of features based on *Root Mean Square Error (RMSQ)* that inform us how far off the model's predicted car prices are from the real prices, on average. That is, the lower the value, the better the model is making predictions. 

#### 5.2.1. RFECV
This method was selected in order to identify an optimal subset of features that maximizes the predictive performance on unseen data. Since *RFECV* starts by analysing all features and recursively removes the least important ones after evaluating the model performance using cross-validation, it is able to determine the optimal number of features that minimizes the validation error.

```python
#combining train and validation datasets
X_combined = np.concatenate([X_train, X_val])
y_combined = np.concatenate([y_train, y_val])

#creating a test fold index (-1 for train, 0 for validation)
test_fold = [-1] * len(X_train) + [0] * len(X_val)

#checking if dimensions are consistent
print('Test fold: ', len(test_fold))
print('X_combined: ', len(X_combined))
print('y_combined: ', len(y_combined))

#defining the PredefinedSplit
ps = PredefinedSplit(test_fold=test_fold)
```

```python
#defining a prepruned random forest regressor

#the hyperparameters limit tree complexity to reduce overfitting
base_estimator = RandomForestRegressor(
    n_estimators=200,        #number of trees in the forest
    max_depth=10,            #maximum depth of each tree
    min_samples_leaf=4,      #minimum samples per leaf
    max_features="sqrt",     #feature sampling strategy
    random_state=0,          #reproducibility
    n_jobs=-1                #using all available CPU cores
)
```

```python
#using negative MSE because scikit-learn expects higher scores to be better
neg_mse_scorer = make_scorer(mean_squared_error, greater_is_better=False)

#executing RFECV to find the optimal feature count based on validation RMSE
#using the balanced estimator for the primary optimization process.
rfecv_selector = RFECV(
    estimator=base_estimator, 
    step=1, #removing one feature at each interaction
    cv=ps,  #using the PredefinedSplit (X_val) for scoring
    scoring=neg_mse_scorer
)
#fitting RFECV on the combined dataset
rfecv_selector.fit(X_combined, y_combined)

#defining the range of features evaluated by RFECV (for the X-axis)
feature_counts_range = rfecv_selector.cv_results_['n_features']

#this loop evaluates how training error evolves as features are removed
train_rmse_list = []
#feature ranking assigned by RFECV (1 = most important)
feature_ranking = rfecv_selector.ranking_

for n in feature_counts_range:
    #selecting features with ranking <= n
    selected_features_mask = feature_ranking <= n
    
    #subseting the training data using the selected features
    X_train_subset = X_train.iloc[:, selected_features_mask]
    
    #training Random Forest on the selected feature subset
    base_estimator.fit(X_train_subset, y_train)
    
    #predicting on the training data
    train_pred = base_estimator.predict(X_train_subset)
    #computing the RMSE on the training set
    train_mse = mean_squared_error(y_train, train_pred)
    train_rmse_list.append(np.sqrt(train_mse))


#optimal number of features chosen by RFECV
optimal_nof = rfecv_selector.n_features_
#retrieving validation negative MSE scores
validation_scores_neg_mse = rfecv_selector.cv_results_['mean_test_score']
#converting negative MSE to positive RMSE
validation_rmse = np.sqrt(-validation_scores_neg_mse) 
#RMSE corresponding to the optimal number of features
min_rmse = validation_rmse[optimal_nof - 1] 

#variables for plotting
nof_list = feature_counts_range
val_rmse_list = validation_rmse
nof = optimal_nof
low_rmse = min_rmse

#displaying the results
print("\n--- RFECV Results (Using Random Forest) ---")
print(f"Optimal number of features found: {nof}")
print(f"Minimum Validation RMSE achieved: {low_rmse:.4f}")

'''This code performs feature selection using RFECV (Recursive Feature Elimination
with Cross-Validation) combined with a Random Forest regressor. The training and 
validation datasets are separated using a PredefinedSplit, ensuring that
feature selection is guided strictly by validation performance.
At each iteration, RFECV removes the least important feature and evaluates the model
using validation RMSE. In parallel, training RMSE is computed manually to analyze
overfitting behavior. '''
```

```python
#plotting training vs validation RMSE
plt.figure(figsize=(10, 6))

#plotting traning RMSE
plt.plot(nof_list, train_rmse_list,
         label="RMSE on Training Set", color='yellowgreen', linewidth=2) 
         
#plotting validation RMSE
plt.plot(nof_list, val_rmse_list, 
         label="RMSE on Validation Set", color='dimgray', linewidth=2) 

#highlighting the optimal feature count
plt.plot(nof, low_rmse, 'ro', 
         label=f'Optimal Features: {nof} (RMSE: {low_rmse:.4f})', markersize=8)

plt.xlabel("Number of Features Selected")
plt.ylabel("RMSE (Root Mean Squared Error)") 
plt.title("RFECV Performance: Training vs. Validation Error")
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

'''The results are visualised to compare training and validation
errors as the number of selected features decreases, allowing identification of the 
optimal feature subset.'''
```

```python
#getting the mask of selected features (True for selected, False otherwise)
selected_feature_mask = rfecv_selector.support_

#getting the names of the selected features
forward_features = X_train.columns[selected_feature_mask]

#printing the results
print("Selected features using RFECV ({} features):".format(optimal_nof))
print(list(forward_features))

'''Outputs the selected features corresponding to the minimum 
validation RMSE.'''
```

According to *RFECV* with *Random Forest* as estimator, the optimal number of features to keep would be 9 as this combination of features reveal the lowest validation *RMSE*. The minimum score was 2938.5113, indicating that the model’s average prediction error is about £2938.
Therefore, according to this method, the variables to be discarded are **previousOwners**, **fuelType_encoded** and **hasDamaged**

#### 5.2.2. Sequential Forward Feature Selector
*SFFS* was also applied to evaluate the relevance of features by building the model through an incremental selection process. 
Unlike elimination-based approaches, *SFS* starts with an empty set and adds the feature that has the greatest improvement in validation performance at each step. This also leads to a optimal group of features that have the lowest validation *RMSE*.

```python
from sklearn.feature_selection import SequentialFeatureSelector

#it starts with no features and adds them
#one by one based on validation performance
sfs_forward = SequentialFeatureSelector(
    estimator=base_estimator,           #model used (random forest regressor) to evaluate feature subsets
    n_features_to_select="auto",        #automatically determine optimal number of features
    direction="forward",                #forward selection strategy
    scoring="neg_mean_squared_error",   #optimization metric (negated for sklearn)
    cv=ps,                              #predefined train/validation split
    n_jobs=-1,                          #parallel computation
    tol=1e-3                            #minimum improvement threshold to continue selection
)

#the model is trained on X_train and evaluated on X_val
#at each feature addition step
sfs_forward.fit(X_combined, y_combined)

#boolean mask indicating what features were selected
forward_mask = sfs_forward.get_support()
#getting the names of the selected features
forward_features = X_train.columns[forward_mask]
#display selected features
print("Forward Selected Features:")
print(list(forward_features))

#fitting the estimator using only the selected features
base_estimator.fit(X_train[forward_features], y_train)

#predicting on the validation set
y_val_pred_forward = base_estimator.predict(X_val[forward_features])
#computing validation RMSE
val_rmse_forward = np.sqrt(mean_squared_error(y_val, y_val_pred_forward))

#predicting on the training set
y_train_pred_forward = base_estimator.predict(X_train[forward_features])
#compute training RMSE
train_rmse_forward = np.sqrt(mean_squared_error(y_train, y_train_pred_forward))

#displaying the results
print("Training RMSE (Forward SFS):", train_rmse_forward)
print("Validation RMSE (Forward SFS):", val_rmse_forward)

'''This code applies Sequential Forward Feature Selection (SFS) to identify 
an optimal group of features for a Random Forest regression model. It starts 
with an empty feature set and then features are added one at a time based 
on how much they improve the validation performance. This performance is
measured using RMSE.
Once the selection process is complete, the model is trained again but this time
using only the selected features and evaluated on both training and validation sets.'''
```

Similiarly to *RFECV*, *Sequential Forward Feature Selection* stated that the optimal number of features would also be 9. When comparing the results of validation *RMSE*, *SFFS* reveals a lower value of less than 50, but with the same subset of features. Moreover, the difference between the training RMSE and the validation RMSE it is not considered significant, which indicates that there no signs of overfitting.
Therefore, according to this method, the features that should be removed are **previousOwners**, **fuelType_encoded** and **hasDamaged**.

```python
def run_benchmarking(X_train, y_train, X_val, y_val):
    #dictionary to store all metrics in a single place
    results = {}

    y_train_true = y_train
    y_val_true   = y_val

    
    #RANDOM FOREST
    rf = RandomForestRegressor(
        n_estimators=200,   
        max_depth=10,
        min_samples_leaf = 4,
        max_features="sqrt",
        random_state=42,
        n_jobs=-1
    )

    #fitting on log-transformed targets
    rf.fit(X_train, y_train_log)

    #converting predictions back to original scale
    y_rf_train = np.expm1(rf.predict(X_train))
    y_rf_val   = np.expm1(rf.predict(X_val))

    #computing train and validation metrics on the original target scale
    results["RF RMSE (Train)"] = np.sqrt(mean_squared_error(y_train_true, y_rf_train))
    results["RF MAE (Train)"]  = mean_absolute_error(y_train_true, y_rf_train)
    results["RF RMSE (Val)"]   = np.sqrt(mean_squared_error(y_val_true, y_rf_val))
    results["RF MAE (Val)"]    = mean_absolute_error(y_val_true, y_rf_val)

    
    #EXTRA TREES
    et = ExtraTreesRegressor(
        n_estimators=400,
        max_depth=12,             
        min_samples_split=20,     
        min_samples_leaf=4,       
        max_features="sqrt",      
        random_state=42,
        n_jobs=-1
    )

    #fitting on log-transformed targets
    et.fit(X_train, y_train_log)

    #converting predictions back to original scale
    y_et_train = np.expm1(et.predict(X_train))
    y_et_val   = np.expm1(et.predict(X_val))

    results["ET RMSE (Train)"] = np.sqrt(mean_squared_error(y_train_true, y_et_train))
    results["ET MAE (Train)"]  = mean_absolute_error(y_train_true, y_et_train)
    results["ET RMSE (Val)"]   = np.sqrt(mean_squared_error(y_val_true, y_et_val))
    results["ET MAE (Val)"]    = mean_absolute_error(y_val_true, y_et_val)


    #MLP 
    mlp = MLPRegressor(
        hidden_layer_sizes=(9, 7),  
        activation="relu",
        solver="adam",
        learning_rate_init=1e-2,
        max_iter=60,
        random_state=42,
        verbose=False
    )

    #fitting on log-transformed targets
    mlp.fit(X_train, y_train_log)

    #converting predictions back to original scale
    y_mlp_train = np.expm1(mlp.predict(X_train))
    y_mlp_val   = np.expm1(mlp.predict(X_val))

    #computing train and validation metrics on the original target scale
    results["MLP RMSE (Train)"] = np.sqrt(mean_squared_error(y_train_true, y_mlp_train))
    results["MLP MAE (Train)"]  = mean_absolute_error(y_train_true, y_mlp_train)
    results["MLP RMSE (Val)"]   = np.sqrt(mean_squared_error(y_val_true, y_mlp_val))
    results["MLP MAE (Val)"]    = mean_absolute_error(y_val_true, y_mlp_val)

    
    #HGBR
    hgbr = HistGradientBoostingRegressor(
        max_iter=100,
        learning_rate=0.1,
        max_depth=None,
        random_state=42,
        l2_regularization=0.0,
        early_stopping=True
    )

    #fittig on log-transformed targets
    hgbr.fit(X_train, y_train_log)

    #converting predictions back to original scale
    y_hgbr_train = np.expm1(hgbr.predict(X_train))
    y_hgbr_val   = np.expm1(hgbr.predict(X_val))

    #computing train and validation metrics on the original target scale
    results["HGBR RMSE (Train)"] = np.sqrt(mean_squared_error(y_train_true, y_hgbr_train))
    results["HGBR MAE (Train)"]  = mean_absolute_error(y_train_true, y_hgbr_train)
    results["HGBR RMSE (Val)"]   = np.sqrt(mean_squared_error(y_val_true, y_hgbr_val))
    results["HGBR MAE (Val)"]    = mean_absolute_error(y_val_true, y_hgbr_val)


    #RIDGE
    #creating Ridge regression model with alpha=1.0
    ridge = Ridge(alpha=1.0)

    #fittig on log-transformed targets
    ridge.fit(X_train, y_train_log)

    #converting predictions back to original scale
    y_ridge_train = np.expm1(ridge.predict(X_train))
    y_ridge_val = np.expm1(ridge.predict(X_val))

    #computing train and validation metrics on the original target scale
    results["Ridge RMSE (Train)"] = np.sqrt(mean_squared_error(y_train_true, y_ridge_train))
    results["Ridge MAE (Train)"]  = mean_absolute_error(y_train_true, y_ridge_train)
    results["Ridge RMSE (Val)"]   = np.sqrt(mean_squared_error(y_val_true, y_ridge_val))
    results["Ridge MAE (Val)"]    = mean_absolute_error(y_val_true, y_ridge_val)


    #LINEAR REGRESSOR
    lr = LinearRegression()

    #fittig on log-transformed targets
    lr.fit(X_train, y_train_log)

    #converting predictions back to original scale
    y_lr_train = np.expm1(lr.predict(X_train))
    y_lr_val = np.expm1(lr.predict(X_val))

    #computing train and validation metrics on the original target scale
    results["LR RMSE (Train)"] = np.sqrt(mean_squared_error(y_train_true, y_lr_train))
    results["LR MAE (Train)"]  = mean_absolute_error(y_train_true, y_lr_train)
    results["LR RMSE (Val)"]   = np.sqrt(mean_squared_error(y_val_true, y_lr_val))
    results["LR MAE (Val)"]    = mean_absolute_error(y_val_true, y_lr_val)

    return pd.DataFrame.from_dict(results, orient="index", columns=["Value"])

results = run_benchmarking( X_train, y_train, X_val, y_val ) 
display(results)

"""
This benchmarking cell compares our regression models 
under the same train/validation split. To make
error metrics interpretable, RMSE and MAE are reported on the original price scale,
even though models are trained on a log-transformed target to reduce skew.
"""
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error
import numpy as np
import pandas as pd
import random

param_dist_mlp = {
    'hidden_layer_sizes': [(9, 9), (9, 8), (8, 9), (9, 9, 9), (9, 8, 8), (8, 8, 9)],
    "learning_rate": [1e-2, 5e-3, 1e-3, 1e-4],
    "max_iter": [200, 300, 500, 800],
    "alpha": [1e-4, 1e-5, 1e-6],
    "solver": ["adam"]
}

def random_search_mlp(param_grid,
                      X_train, y_train,
                      X_val,   y_val,
                      n_iter=60):

    results = []
    best_rmse_val = float("inf")
    best_model = None
    best_params = None

    # original scale
    y_train_true = y_train
    y_val_true   = y_val

    keys = list(param_grid.keys())

    for i in range(n_iter):
        # random pick from parameter grid
        params = {k: random.choice(param_grid[k]) for k in keys}

        mlp = MLPRegressor(
            hidden_layer_sizes=params["hidden_layer_sizes"],
            activation="relu",
            solver=params["solver"],           
            learning_rate_init=params["learning_rate"],
            max_iter=params["max_iter"],
            alpha=params["alpha"],            
            random_state=42,
            verbose=False
        )

        # train on log-transformed target
        mlp.fit(X_train, y_train_log)

        # predict with log
        pred_train_log = mlp.predict(X_train)
        pred_val_log   = mlp.predict(X_val)

        # back to original scale
        pred_train = np.expm1(pred_train_log)
        pred_val   = np.expm1(pred_val_log)

        rmse_train = mean_squared_error(y_train_true, pred_train) ** 0.5
        rmse_val   = mean_squared_error(y_val_true,   pred_val) ** 0.5

        results.append({
            **params,
            "rmse_train": rmse_train,
            "rmse_val": rmse_val
        })

        print(f"[{i+1}/{n_iter}] rmse_train={rmse_train:.2f} | rmse_val={rmse_val:.2f}")

        # keep best model
        if rmse_val < best_rmse_val:
            best_rmse_val = rmse_val
            best_model = mlp
            best_params = params

    results_mlp_df = pd.DataFrame(results).sort_values("rmse_val").reset_index(drop=True)

    return results_mlp_df, best_model, best_params

# Call the function with raw targets
results_mlp_df, best_mlp_model, best_mlp_params = random_search_mlp(
    param_dist_mlp,
    X_train, y_train,
    X_val,   y_val,
    n_iter=60  
)

print("\nBest parameters:")
print(best_mlp_params)
print(f"Best RMSE_val MLP: {results_mlp_df['rmse_val'].iloc[0]:.2f}")
```

### 6.2. Tuned Benchmarking

```python
def final_benchmark_log(
    X_train, y_train_log,
    X_val,   y_val_log,
    rf_params,
    et_params,
    mlp_params,     
    hgbr_params,
    ridge_params
):
    results = {}
    fitted_models = {}

    # ======================
    # Random Forest
    # ======================
    rf = RandomForestRegressor(
        random_state=42,
        n_jobs=-1,
        **rf_params
    )
    rf.fit(X_train, y_train_log)
    fitted_models["rf"] = rf

    pred_train = np.expm1(rf.predict(X_train))
    pred_val   = np.expm1(rf.predict(X_val))

    results["RF_RMSE_train"] = mean_squared_error(y_train, pred_train) ** 0.5
    results["RF_MAE_train"]  = mean_absolute_error(y_train, pred_train)
    results["RF_RMSE_val"]   = mean_squared_error(y_val, pred_val) ** 0.5
    results["RF_MAE_val"]    = mean_absolute_error(y_val, pred_val)

    # ======================
    # Extra Trees
    # ======================
    et = ExtraTreesRegressor(
        random_state=42,
        n_jobs=-1,
        **et_params
    )
    et.fit(X_train, y_train_log)
    fitted_models["et"] = et

    pred_train = np.expm1(et.predict(X_train))
    pred_val   = np.expm1(et.predict(X_val))

    results["ET_RMSE_train"] = mean_squared_error(y_train, pred_train) ** 0.5
    results["ET_MAE_train"]  = mean_absolute_error(y_train, pred_train)
    results["ET_RMSE_val"]   = mean_squared_error(y_val, pred_val) ** 0.5
    results["ET_MAE_val"]    = mean_absolute_error(y_val, pred_val)

    # ======================
    # MLP (sklearn replacement for DNN)
    # ======================
    mlp = MLPRegressor(
        hidden_layer_sizes=mlp_params["hidden_layer_sizes"],
        activation="relu",
        solver=mlp_params["solver"],
        learning_rate_init=mlp_params["learning_rate"],
        max_iter=mlp_params["max_iter"],
        alpha=mlp_params["alpha"],
        random_state=42,
        verbose=False
    )
    mlp.fit(X_train, y_train_log)
    fitted_models["mlp"] = mlp

    pred_train = np.expm1(mlp.predict(X_train))
    pred_val   = np.expm1(mlp.predict(X_val))

    results["MLP_RMSE_train"] = mean_squared_error(y_train, pred_train) ** 0.5
    results["MLP_MAE_train"]  = mean_absolute_error(y_train, pred_train)
    results["MLP_RMSE_val"]   = mean_squared_error(y_val, pred_val) ** 0.5
    results["MLP_MAE_val"]    = mean_absolute_error(y_val, pred_val)

    # ======================
    # HGBR
    # ======================
    hgbr = HistGradientBoostingRegressor(
        random_state=42,
        **hgbr_params
    )
    hgbr.fit(X_train, y_train_log)
    fitted_models["hgbr"] = hgbr

    pred_train = np.expm1(hgbr.predict(X_train))
    pred_val   = np.expm1(hgbr.predict(X_val))

    results["HGBR_RMSE_train"] = mean_squared_error(y_train, pred_train) ** 0.5
    results["HGBR_MAE_train"]  = mean_absolute_error(y_train, pred_train)
    results["HGBR_RMSE_val"]   = mean_squared_error(y_val, pred_val) ** 0.5
    results["HGBR_MAE_val"]    = mean_absolute_error(y_val, pred_val)

    # ======================
    # Ridge Regressor
    # =====================
    ridge = Ridge(
        **ridge_params,
        random_state=42
    )
    ridge.fit(X_train, y_train_log)
    fitted_models["ridge"] = ridge

    pred_train = np.expm1(ridge.predict(X_train))
    pred_val   = np.expm1(ridge.predict(X_val))

    results["Ridge_RMSE_train"] = mean_squared_error(y_train, pred_train) ** 0.5
    results["Ridge_MAE_train"]  = mean_absolute_error(y_train, pred_train)
    results["Ridge_RMSE_val"]   = mean_squared_error(y_val, pred_val) ** 0.5
    results["Ridge_MAE_val"]    = mean_absolute_error(y_val, pred_val)

    # ======================
    # Linear Regression
    # ======================
    lr = LinearRegression()
    lr.fit(X_train, y_train_log)
    fitted_models["linear"] = lr

    pred_train = np.expm1(lr.predict(X_train))
    pred_val   = np.expm1(lr.predict(X_val))

    results["LR_RMSE_train"] = mean_squared_error(y_train, pred_train) ** 0.5
    results["LR_MAE_train"]  = mean_absolute_error(y_train, pred_train)
    results["LR_RMSE_val"]   = mean_squared_error(y_val, pred_val) ** 0.5
    results["LR_MAE_val"]    = mean_absolute_error(y_val, pred_val)

    # ======================
    # Results table
    # ======================
    df = pd.DataFrame([
        ["RandomForest", results["RF_RMSE_train"], results["RF_MAE_train"],
         results["RF_RMSE_val"], results["RF_MAE_val"]],
        ["ExtraTrees", results["ET_RMSE_train"], results["ET_MAE_train"],
         results["ET_RMSE_val"], results["ET_MAE_val"]],
        ["MLP", results["MLP_RMSE_train"], results["MLP_MAE_train"],
         results["MLP_RMSE_val"], results["MLP_MAE_val"]],
        ["HGBR", results["HGBR_RMSE_train"], results["HGBR_MAE_train"],
         results["HGBR_RMSE_val"], results["HGBR_MAE_val"]],
        ["Ridge", results["Ridge_RMSE_train"], results["Ridge_MAE_train"],
         results["Ridge_RMSE_val"], results["Ridge_MAE_val"]],
        ["Linear", results["LR_RMSE_train"], results["LR_MAE_train"],
         results["LR_RMSE_val"], results["LR_MAE_val"]],
    ], columns=["Model", "RMSE_train", "MAE_train", "RMSE_val", "MAE_val"])

    return df.set_index("Model"), fitted_models
```

```python
# Best hyperparameters from random search

rf_best_params = {
    "n_estimators": 500,
    "max_depth": 15,
    "min_samples_leaf": 8,
    "max_features": 0.5,
}

et_best_params = {
    "n_estimators": 300,
    "min_samples_leaf": 2,
    "min_samples_split": 20,
    "max_depth": None,
    "max_features": 0.8,
} 

mlp_best_params = {
    "hidden_layer_sizes": (9,9,9),
    "learning_rate": 0.005,
    "max_iter": 300,
    "alpha": 0.0001,
    "solver": "adam"
}


hgbr_params = {
    'learning_rate': 0.05, 
    'max_iter': 1000, 
    'max_leaf_nodes': 63, 
    'min_samples_leaf': 10, 
    'l2_regularization': 1.0, 
    'max_depth': None, 
    'early_stopping': False
}

ridge_params = {
    "alpha": 10,   
    "solver": "auto" 
}

benchmark_log_df, final_models = final_benchmark_log(
    X_train, y_train_log,
    X_val,   y_val_log,
    rf_params=rf_best_params,
    et_params=et_best_params,
    mlp_params=mlp_best_params,
    hgbr_params=hgbr_params,
    ridge_params=ridge_params
)

display(benchmark_log_df)
```

After the individual optimization of each model, we created a benchmarking in order to analyse the performance of each model with the chosen hyperparamethers. Based on these results, we decided to test a stacking experimental model, so that we could evaluate if the combination of the models could improve their performance, comparing with the original result.

Com base nos resultados do benchmarking final, apenas os modelos com desempenho competitivo no conjunto de validação foram considerados para o stacking. O modelo DNN foi excluído por apresentar um erro significativamente superior, o que poderia introduzir ruído no modelo ensemble. O stacking foi assim construído com os modelos HGBR e ExtraTrees, que apresentaram bom desempenho individual e diferentes mecanismos de aprendizagem

#### 6.2.1. Stacking

```python
def rmse(y_true, y_pred):
    return mean_squared_error(y_true, y_pred) ** 0.5

# Final models
hgbr = final_models["hgbr"]
et   = final_models["et"]

# Base predictions
hgbr_val_log = hgbr.predict(X_val)
et_val_log   = et.predict(X_val)

y_train_log = np.log1p(y_train)
y_val_log   = np.log1p(y_val)

X_meta_val = np.column_stack([hgbr_val_log, et_val_log])
y_meta_val = y_val_log

# Meta-model
meta_model = Ridge(alpha=1.0, random_state=42)
meta_model.fit(X_meta_val, y_meta_val)

# Real scale evaluation
y_val_true = np.expm1(y_val_log)

hgbr_val_pred  = np.expm1(hgbr_val_log)
et_val_pred    = np.expm1(et_val_log)
stack_val_pred = np.expm1(meta_model.predict(X_meta_val))

stack_results = pd.DataFrame({
    "Model": ["HGBR", "ExtraTrees", "STACK (HGBR + ET)"],
    "RMSE_val": [
        rmse(y_val_true, hgbr_val_pred),
        rmse(y_val_true, et_val_pred),
        rmse(y_val_true, stack_val_pred),
    ],
    "MAE_val": [
        mean_absolute_error(y_val_true, hgbr_val_pred),
        mean_absolute_error(y_val_true, et_val_pred),
        mean_absolute_error(y_val_true, stack_val_pred),
    ]
}).set_index("Model")

display(stack_results)

# saving the stacking meta-model
final_models["stack_meta_hgbr_et"] = meta_model
```

A stacking ensemble was built using the two best-performing individual models (HistGradientBoostingRegressor and ExtraTreesRegressor).
The base models were trained on the logarithm of the target variable.
Their validation predictions were then used as inputs to a Ridge regression meta-model (α = 1.0), also trained in log-space.
Final predictions were transformed back to the original scale for evaluation using RMSE and MAE

The stacking ensemble achieved a validation performance comparable to the best individual model (HGBR), confirming its robustness. Although no significant improvement was observed, the stacking model did not degrade performance and slightly improved MAE, suggesting a stabilizing effect rather than a performance boost

Three versions of the project have been created using different preprocessing approaches, namely handling categorical variables and imputing missing values. The first version corresponds to the current project, which uses *Frequency Encoder* to transform categorical variables into numerical ones and an *Iterative Imputer* to handle missing values. Two additional versions were developed: one using a *KNN Imputer* for missing value imputation combined with a *Frequency Encoder* for all categorical variables, and another using a KNN Imputer while applying *One Hot Encoder* to the variables *fuelType* and *transmission*, with the remaining categorical variables encoded using a *Frequency Encoder*.
